### Step 1: Extract S&P 500 Tickers

I will use `pandas.read_html` to extract the S&P 500 component list from Wikipedia. This requires the `lxml` library, so I'll install it first if it's not already available.

```python
# Install required packages
!pip install yfinance yahoofinancials lxml html5lib --quiet

import pandas as pd

url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

# Add User-Agent header to mimic a browser request
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}
tables = pd.read_html(url, storage_options=headers)

df = tables[0]  # first table on the page
df.head()
```

### Step 1.1: Save S&P 500 Tickers to CSV

Now that we have extracted the S&P 500 list into a DataFrame, I will save it to a CSV file named `sp500_tickers.csv`.

```python
# Save the DataFrame to a CSV file
df.to_csv('sp500_tickers.csv', index=False)

print("S&P 500 tickers saved to 'sp500_tickers.csv'.")

# Display the first few rows of the DataFrame to confirm
display(df.head())
```

### Step 2: Load S&P 500 Tickers Daily Market Performance

I will use the `yfinance` package to download the daily market performance data for the S&P 500 tickers for the period 01-01-2022 to 01-01-2025. I'll first extract the 'Symbol' column from our `df` to get the list of tickers.

```python
import yfinance as yf

# Extract the 'Symbol' column to get the list of tickers
sp500_tickers = df['Symbol'].tolist()

start_date = '2022-01-01'
end_date = '2025-01-01'

print(f"Downloading daily market performance for {len(sp500_tickers)} tickers from {start_date} to {end_date}...")

# Download data for all S&P 500 tickers
data = yf.download(sp500_tickers, start=start_date, end=end_date)

print("Data downloaded successfully. Displaying the first 5 rows of the 'Adj Close' prices:")
display(data['Adj Close'].head())
```
