Now, you will use `git add`, `git commit`, and `git push` to upload your file to the GitHub repository. You will need to provide your GitHub username and a Personal Access Token (PAT) when prompted for credentials by `git push`.

```python
# Add the new file to the staging area
!git add Group_Coursework
print("Staged 'Group_Coursework' for commit.")

# Configure Git user identity (required for committing)
# IMPORTANT: Replace with your GitHub email and username
!git config user.email "your_email@example.com" # <--- PLEASE REPLACE THIS WITH YOUR EMAIL
!git config user.name "Your Name" # <--- PLEASE REPLACE THIS WITH YOUR NAME
print("Set Git user identity. Please remember to replace 'your_email@example.com' and 'Your Name' with your actual GitHub credentials.")

# Commit the changes
commit_message = "Add Group_Coursework file"
!git commit -m "{commit_message}"
print(f"Committed with message: '{commit_message}'")

# Push the changes to the remote repository (main branch is common, but could be 'master')
# You will be prompted for your GitHub username and Personal Access Token (PAT).
# If the interactive prompt does not appear or fails (e.g., 'No such device or address' error),
# you may need to generate a GitHub Personal Access Token (PAT) and embed it in the URL.
# Example of using PAT in the URL (replace YOUR_GITHUB_TOKEN and ensure your PAT has 'repo' scope):
!git push https://YOUR_GITHUB_TOKEN@github.com/naumankhokhar/sp500-ml-clustering-portfolio-optimization.git main # <--- UNCOMMENT AND USE THIS LINE WITH YOUR PAT, THEN REPLACE 'YOUR_GITHUB_TOKEN'
# !git push origin main # <--- THIS LINE WILL LIKELY FAIL WITHOUT PAT/INTERACTIVE AUTH
print("Attempted to push changes to GitHub. Please check your repository online for successful synchronization.")
```

### Managing Group Work on GitHub following your structure

Now that you know how to push changes, let's discuss how to effectively manage group work based on your repository structure (`/data`, `/code`, `/reports`, `/docs`).

**1. Branching Strategy:**
It's highly recommended that each team member works on a separate branch for their features or tasks. A common workflow is:
*   **`main` (or `master`) branch**: This branch should always contain stable, production-ready code. No direct commits should be made here.
*   **`development` branch**: All new features or work are merged into this branch. This is your integration branch.
*   **Feature branches**: Each team member creates a new branch off `development` for their specific task (e.g., `feature/data_preprocessing`, `feature/k_means_clustering`, `bugfix/report_bug`).

**Workflow Example:**
1.  **Pull latest `development`**: `git pull origin development`
2.  **Create new feature branch**: `git checkout -b feature/my-task development`
3.  **Work on your task**: Make changes to files within your assigned directories (e.g., `/code`, `/data`).
4.  **Add and Commit**: `git add .` then `git commit -m "Implement my task"`
5.  **Push your feature branch**: `git push origin feature/my-task`
6.  **Create a Pull Request (PR)**: Go to GitHub, compare your `feature/my-task` branch with `development`, and create a PR.
7.  **Code Review**: Team members review the PR, provide feedback, and request changes.
8.  **Merge**: Once approved, merge the PR into `development`.
9.  **Periodically sync `development` with `main`**: When `development` is stable, merge it into `main` (usually via a PR) and tag a release.

**2. Repository Structure Utilization:**
*   **`/data`**: This directory should contain datasets used by the project. It's often best practice to store relatively small datasets directly or use Git LFS for larger ones. Avoid committing sensitive data.
*   **`/code`**: This is where your Python scripts and Jupyter notebooks will live. Organize them logically (e.g., `code/notebooks`, `code/scripts`, `code/modules`).
*   **`/reports`**: Use this for final reports, presentations, or generated analysis summaries.
*   **`/docs`**: Store any supporting documentation, project plans, meeting notes, or visual assets here.

**3. Collaboration Tools:**
*   **GitHub Issues**: Use issues to track tasks, bugs, and feature requests. Assign issues to team members.
*   **Pull Requests (PRs)**: Essential for code review and ensuring quality before merging.
*   **Project Boards**: Use GitHub Projects to visualize your workflow (e.g., To Do, In Progress, Done).

By following these practices, your team can collaborate efficiently and maintain a clean, organized repository.

```python
import requests

# Construct the raw content URL for the README.md file
# The base URL for raw GitHub content is different from the blob URL
repo_owner = "naumankhokhar"
repo_name = "sp500-ml-clustering-portfolio-optimization"
branch = "main"
file_path = "README.md"

raw_url = f"https://raw.githubusercontent.com/{repo_owner}/{repo_name}/{branch}/{file_path}"

print(f"Fetching content from: {raw_url}")

try:
    response = requests.get(raw_url)
    response.raise_for_status()  # Raise an exception for HTTP errors (4xx or 5xx)
    readme_content = response.text
    print("\n--- Content of README.md ---\n")
    print(readme_content)
    print("\n--------------------------")
except requests.exceptions.RequestException as e:
    print(f"Error fetching the file: {e}")
```
